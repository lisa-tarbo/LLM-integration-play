```python
# https://ai.google.dev/gemini-api/docs/gemini-3
import os
from dotenv import load_dotenv
from google import genai
from google.api_core.exceptions import GoogleAPICallError

load_dotenv()
api_key = os.getenv("GOOGLE_API_KEY")

if not api_key:
    raise ValueError("GOOGLE_API_KEY environment variable not set")

client = genai.Client(api_key=api_key)

try:
    response = client.models.generate_content(
        model="gemini-3.1-pro-preview",
        contents="When is next year's World Cup?",
    )

    print(response.text)

except GoogleAPICallError as exc:
    print("API request failed:", exc)
```

```python
import os
from dotenv import load_dotenv
from google import genai
from google.genai import types

load_dotenv()
api_key = os.getenv("GOOGLE_API_KEY")

if not api_key:
    raise ValueError("GOOGLE_API_KEY environment variable not set")

client = genai.Client(api_key=api_key)

response = client.models.generate_content(
    model="gemini-3.1-pro-preview",
    contents="How does AI work? In 2 sentences.",
    config=types.GenerateContentConfig(
        thinking_config=types.ThinkingConfig(thinking_level="low")
    ),
)

print(response.text)
```

```python
# See text response and metadata response

# https://docs.langchain.com/oss/python/integrations/chat/google_generative_ai

import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
api_key = os.getenv("GOOGLE_API_KEY")

if not api_key:
    raise ValueError("GOOGLE_API_KEY environment variable not set")

model = ChatGoogleGenerativeAI(model="gemini-3.1-pro-preview", api_key=api_key)
llm_response = model.invoke("Write me a ballad about LangChain in 1 sentence")

print(llm_response.text)
print("\n\n")
print(llm_response.response_metadata)
```
