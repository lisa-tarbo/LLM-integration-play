## OpenAI — Responses API vs. Chat Completions

This notebook favors the newer Responses API over the older Chat Completions API. The Responses API is an agentic loop: it lets the model call tools such as `web_search`, `image_generation`, `file_search`, and custom functions.

### API key check
The remaining cells assume `OPENAI_API_KEY` is already set as an environment variable.

```python
from openai import OpenAI
import os
from dotenv import load_dotenv

load_dotenv()

api_key = os.getenv("OPENAI_API_KEY")

if not api_key:
    raise ValueError("OPENAI_API_KEY environment variable not set")
```

### Basic Responses API call

Minimal `responses.create` call, with `AuthenticationError` handling for a missing or invalid key. 

```python
from openai import OpenAI, AuthenticationError

client = OpenAI()

try:
    response = client.responses.create(
        model="gpt-5-nano", input="Write a one-sentence bedtime story about a unicorn."
    )
    print(response.output_text)
except AuthenticationError:
    print(
        "Authentication failed (401): please check that your OPENAI_API_KEY is set and valid."
    )
```

### `instructions` + `reasoning` parameters

Uses the `instructions` field (a system-style prompt) plus `reasoning.effort` to steer response style and reasoning depth.

```python
from openai import OpenAI

client = OpenAI()

response = client.responses.create(
    model="gpt-5-nano",
    reasoning={"effort": "low"},
    instructions="Talk like a pirate.",
    input="Are semicolons optional in JavaScript?",
)

print(response.output_text)
```

### Same call via a role-based `input` array

Equivalent to the previous cell, but passes the system-style prompt as a `developer`-role message in the `input` array instead of using `instructions`.

```python
from openai import OpenAI

client = OpenAI()

response = client.responses.create(
    model="gpt-5",
    reasoning={"effort": "low"},
    input=[
        {"role": "developer", "content": "Talk like a pirate."},
        {"role": "user", "content": "Are semicolons optional in JavaScript?"},
    ],
)

print(response.output_text)
```

### Legacy Chat Completions API

For comparison: the older `chat.completions.create` interface, kept as a reference alongside the Responses API examples above.
Reference: https://developers.openai.com/api/docs/guides/migrate-to-responses

```python
from openai import OpenAI

client = OpenAI()

completion = client.chat.completions.create(
    model="gpt-4o-mini-2024-07-18",
    messages=[
        {"role": "user", "content": "Write a one-sentence bedtime story about an owl."}
    ],
)

print(completion.choices[0].message.content)
```
