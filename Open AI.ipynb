```python
import os
from langchain_openai.chat_models import ChatOpenAI
from langchain.schema import HumanMessage

def count_tokens(prompt: str, model: str = "gpt-4o-mini") -> int:
    """
    return the number of tokens that `model` would consume for a single
    user message.
    """
    # make sure the key is available; the LangChain client will read
    # from OPENAI_API_KEY by default.
    if "OPENAI_API_KEY" not in os.environ:
        raise RuntimeError("set OPENAI_API_KEY in the environment")

    llm = ChatOpenAI(model=model, temperature=0)
    messages = [HumanMessage(content=prompt)]
    return llm.get_num_tokens_from_messages(messages)


if __name__ == "__main__":
    prompt = "Hello, how many tokens does this use?"
    print(f"prompt: {prompt!r}")
    print("token count:", count_tokens(prompt))
```
